```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load the dataset created in Task 1
df = pd.read_csv('iris.csv')

# 1. Handle missing data (Removal)
# The standard Iris dataset usually doesn't have nulls, but this is the required logic.
df.dropna(inplace=True)

# 2. Detect and remove outliers
# Using the IQR (Interquartile Range) method for the numerical columns
numerical_cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    # Filtering rows that are within 1.5 * IQR
    df = df[~((df[col] < (Q1 - 1.5 * IQR)) | (df[col] > (Q3 + 1.5 * IQR)))]

# 3. Convert categorical variables into numerical format (Label Encoding)
# This converts species names (Setosa, etc.) into 0, 1, 2
le = LabelEncoder()
df['species'] = le.fit_transform(df['species'])

# 4. Normalize or standardize numerical data
# Scaling the features so they have a mean of 0 and variance of 1
scaler = StandardScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

# Final check
print("Preprocessed Data Head:")
print(df.head())
```
